#Red Densa

```
import numpy as np
import tensorflow as tf
(x_train, y_train), (x_test, y_test)=tf.keras.datasets.mnist.load_data()
x_train=x_train.reshape(x_train.shape[0], 28*28)
x_test=x_test.reshape(x_test.shape[0], 28*28)
import matplotlib.pyplot as plt
plt.imshow(x_train[0].reshape(28,28), cmap='gray')
```

```
In=tf.keras.layers.Input(shape=(784,))
Cp1=tf.keras.layers.Dense(64,activation='relu')(In)
Cp2=tf.keras.layers.Dense(64,activation='relu')(Cp1)
Out=tf.keras.layers.Dense(10,activation='softmax')(Cp2)
Model=tf.keras.Model(inputs=In,outputs=Out)
```

```
Model.summary()
```

```
Model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
Model.fit(x_train,y_train,epochs=30,batch_size=32)
```

#Convolucional

```
import numpy as np
import tensorflow as tf
(x_train, y_train), (x_test, y_test)=tf.keras.datasets.mnist.load_data()

x_train = x_train[..., np.newaxis]

print("Shape de x_train:", x_train.shape)
```

```
In=tf.keras.layers.Input(shape=(28,28,1))
conv1=tf.keras.layers.Conv2D(32,kernel_size=(3,3),activation='relu')(In)
conv2=tf.keras.layers.Conv2D(64,kernel_size=(3,3),activation='relu')(conv1)
f=tf.keras.layers.Flatten()(conv2)
Out=tf.keras.layers.Dense(10,activation='softmax')(f)
Model=tf.keras.Model(inputs=In,outputs=Out)
```

```
Model.summary()
```

```
Model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
Model.fit(x_train,y_train,epochs=30,batch_size=32)
```
