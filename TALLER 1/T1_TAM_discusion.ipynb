## Discusión sobre la Elección de $k$ en k-NN

* **Impacto del tamaño de $k$**:
  * **$k$ muy pequeño ($k=1$)**: El modelo se vuelve extremadamente flexible y sensible al ruido. Genera fronteras complejas que memorizan los datos de entrenamiento (Accuracy de train = 100%), incurriendo en sobreajuste (*overfitting*).
  * **$k$ muy grande (e.g. $k=15$ o cercano a $N$)**: La frontera se vuelve excesivamente rígida y suave. El modelo simplifica en exceso la estructura subyacente y tiende al subajuste (*underfitting*), guiándose únicamente por la clase mayoritaria global.

* **Selección del mejor modelo**:
  * **$k=5$** obtuvo el mejor desempeño con una exactitud en prueba del 95.56% (frente a 93.33% de $k=1$ y $k=15$). Mantiene un equilibrio óptimo entre el sesgo y la varianza del modelo.

* **¿Por qué seleccionar $k$ con Validación Cruzada y no con el Test Set?**:
  * El conjunto de prueba (*test set*) debe ser mantenido estricta e independientemente para la evaluación final imparcial del sistema. Si los hiperparámetros como $k$ se sintonizan maximizando el rendimiento sobre los datos de prueba, se incurre en fuga de información (*data leakage*) y sesgo de optimismo, invalidando al conjunto de prueba como métrica externa real. La optimización formal debe realizarse usando validación cruzada ($K$-Fold CV) sobre el conjunto de entrenamiento.

## Situaciones donde la Exactitud (Accuracy) resulta engañosa

1. **Conjuntos de Datos Desbalanceados**:
   En problemas donde una clase abarca la inmensa mayoría del conjunto (por ejemplo, detección de fraudes donde el 99.5% de transacciones son legítimas), un clasificador ingenuo que prediga siempre "no fraude" obtendrá un 99.5% de Accuracy sin haber detectado un solo caso positivo real ($\text{Recall} = 0$).

2. **Costos Asimétricos de Error (Falsos Positivos vs. Falsos Negativos)**:
   En el diagnóstico médico crítico (por ejemplo, detección de tumores malignos), cometer un Falso Negativo (no diagnosticar a un enfermo) tiene un costo de vida potencialmente fatal, mientras que un Falso Positivo solo implica realizar exámenes adicionales. La exactitud evalúa ambos errores por igual, ocultando el riesgo del modelo en escenarios de alto impacto.
